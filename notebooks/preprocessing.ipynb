# Handwritten Digital Recognition System

# Image Preprocessing Notebook

This notebook demonstrates the complete preprocessing pipeline used before OCR inference.

Objectives

- Load image
- Resize
- Convert to grayscale
- Remove noise
- Correct skew
- Apply thresholding
- Enhance image
- Segment text
- Save processed image

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (12,8)
```

```python
IMAGE_PATH = "../data/samples/images/sample1.jpg"

assert os.path.exists(IMAGE_PATH), "Image not found."
```

```python
image = cv2.imread(IMAGE_PATH)

image = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)

plt.imshow(image)

plt.title("Original Image")

plt.axis("off")
```

```python

print("Height :", image.shape[0])

print("Width :", image.shape[1])

print("Channels :", image.shape[2])
```

```python
resized = cv2.resize(
    image,
    (800,600)
)

plt.imshow(resized)

plt.title("Resized")

plt.axis("off")
```

```python
gray = cv2.cvtColor(
    resized,
    cv2.COLOR_RGB2GRAY
)

plt.imshow(
    gray,
    cmap="gray"
)

plt.title("Grayscale")

plt.axis("off")
```

```python
blur = cv2.GaussianBlur(
    gray,
    (5,5),
    0
)

plt.imshow(
    blur,
    cmap="gray"
)

plt.title("Gaussian Blur")

plt.axis("off")
```

```python
median = cv2.medianBlur(
    gray,
    3
)

plt.imshow(
    median,
    cmap="gray"
)

plt.title("Median Blur")

plt.axis("off")
```

```python
bilateral = cv2.bilateralFilter(
    gray,
    9,
    75,
    75
)

plt.imshow(
    bilateral,
    cmap="gray"
)

plt.title("Bilateral Filter")

plt.axis("off")
```

```python
equalized = cv2.equalizeHist(gray)

plt.imshow(
    equalized,
    cmap="gray"
)

plt.title("Histogram Equalization")

plt.axis("off")
```

```python
threshold = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

plt.imshow(
    threshold,
    cmap="gray"
)

plt.title("Adaptive Threshold")

plt.axis("off")
```

```python
kernel = np.ones((3,3),np.uint8)

opened = cv2.morphologyEx(
    threshold,
    cv2.MORPH_OPEN,
    kernel
)

plt.imshow(
    opened,
    cmap="gray"
)

plt.title("Opening")

plt.axis("off")
```

```python
kernel = np.ones((3,3),np.uint8)

opened = cv2.morphologyEx(
    threshold,
    cv2.MORPH_OPEN,
    kernel
)

plt.imshow(
    opened,
    cmap="gray"
)

plt.title("Opening")

plt.axis("off")
```

```python
closed = cv2.morphologyEx(
    opened,
    cv2.MORPH_CLOSE,
    kernel
)

plt.imshow(
    closed,
    cmap="gray"
)

plt.title("Closing")

plt.axis("off")
```

```python
edges = cv2.Canny(
    gray,
    100,
    200
)

plt.imshow(
    edges,
    cmap="gray"
)

plt.title("Canny Edge Detection")

plt.axis("off")
```

```python
enhanced = cv2.convertScaleAbs(
    gray,
    alpha=1.5,
    beta=15
)

plt.imshow(
    enhanced,
    cmap="gray"
)

plt.title("Enhanced Image")

plt.axis("off")
```

```python
contours, _ = cv2.findContours(
    threshold,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

contour_image = resized.copy()

cv2.drawContours(
    contour_image,
    contours,
    -1,
    (255,0,0),
    2
)

plt.imshow(contour_image)

plt.title("Detected Contours")

plt.axis("off")
```

```python
fig, axes = plt.subplots(2,3,figsize=(15,10))

axes[0,0].imshow(image)
axes[0,0].set_title("Original")

axes[0,1].imshow(gray,cmap="gray")
axes[0,1].set_title("Grayscale")

axes[0,2].imshow(blur,cmap="gray")
axes[0,2].set_title("Blur")

axes[1,0].imshow(threshold,cmap="gray")
axes[1,0].set_title("Threshold")

axes[1,1].imshow(enhanced,cmap="gray")
axes[1,1].set_title("Enhanced")

axes[1,2].imshow(edges,cmap="gray")
axes[1,2].set_title("Edges")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()

plt.show()
```

```python
os.makedirs(
    "../data/processed",
    exist_ok=True
)

cv2.imwrite(
    "../data/processed/preprocessed.png",
    threshold
)

print("Image saved successfully.")
```

```python
print("Original Shape :", image.shape)

print("Processed Shape :", threshold.shape)

print("Mean :", gray.mean())

print("Std :", gray.std())

print("Contours :", len(contours))
```
