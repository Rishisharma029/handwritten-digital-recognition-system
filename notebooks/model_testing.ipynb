# Evaluation Summary

This notebook evaluated:

- OCR Accuracy
- Character Error Rate (CER)
- Word Error Rate (WER)
- Confidence Scores
- Processing Time
- OCR Engine Comparison

The evaluation results can be used to:

- Compare OCR engines
- Benchmark preprocessing methods
- Select the best model for deployment
- Monitor OCR quality over time

```python
import os
import time
import cv2
import matplotlib.pyplot as plt
from PIL import Image

from backend.ocr.easyocr_engine import EasyOCREngine
from backend.ocr.trocr_engine import TrOCREngine
from backend.ocr.paddle_engine import PaddleOCREngine
from backend.ocr.tesseract_engine import TesseractEngine
```

```python
IMAGE_PATH = "../data/samples/images/sample1.jpg"

assert os.path.exists(IMAGE_PATH), "Image not found."
```

```python
image = cv2.imread(IMAGE_PATH)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10,6))
plt.imshow(image)
plt.axis("off")
plt.title("Test Image")
```

```python
easyocr_engine = EasyOCREngine()

trocr_engine = TrOCREngine()

paddle_engine = PaddleOCREngine()

tesseract_engine = TesseractEngine()
```

```python
def test_engine(name, engine, image_path):

    start = time.time()

    result = engine.recognize(image_path)

    end = time.time()

    return {
        "engine": name,
        "text": result["text"],
        "confidence": result["confidence"],
        "processing_time": round(end-start,3)
    }
```

```python
easy_result = test_engine(
    "EasyOCR",
    easyocr_engine,
    IMAGE_PATH
)

easy_result
```

```python
trocr_result = test_engine(
    "TrOCR",
    trocr_engine,
    IMAGE_PATH
)

trocr_result
```

```python
paddle_result = test_engine(
    "PaddleOCR",
    paddle_engine,
    IMAGE_PATH
)

paddle_result
```

```python
tesseract_result = test_engine(
    "Tesseract",
    tesseract_engine,
    IMAGE_PATH
)

tesseract_result
```

```python
import pandas as pd

results = pd.DataFrame([
    easy_result,
    trocr_result,
    paddle_result,
    tesseract_result
])

results
```

```python
results.plot.bar(
    x="engine",
    y="confidence",
    legend=False,
    figsize=(8,5)
)

plt.ylabel("Confidence")
plt.title("OCR Confidence Comparison")
plt.show()
```

```python
results.plot.bar(
    x="engine",
    y="processing_time",
    legend=False,
    figsize=(8,5)
)

plt.ylabel("Seconds")
plt.title("Inference Time Comparison")
plt.show()
```

```python
best = results.loc[
    results["confidence"].idxmax()
]

best
```

```python
for _, row in results.iterrows():

    print("="*60)

    print(row["engine"])

    print()

    print(row["text"])

    print()

    print("Confidence:", row["confidence"])
```

```python
results.to_csv(
    "../outputs/model_testing.csv",
    index=False
)

print("Results saved.")
```

```python
results.describe()
```

```python
print(
    "Highest Confidence:",
    results.loc[
        results["confidence"].idxmax(),
        "engine"
    ]
)

print(
    "Fastest Model:",
    results.loc[
        results["processing_time"].idxmin(),
        "engine"
    ]
)
```

```python
results.to_json(
    "../outputs/model_testing.json",
    orient="records",
    indent=4
)

print("JSON exported.")
```
