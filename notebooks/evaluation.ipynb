# Handwritten Digital Recognition System

# OCR Model Evaluation

This notebook evaluates OCR engine performance on handwritten datasets.

Supported OCR Engines:

- EasyOCR
- TrOCR
- PaddleOCR
- Tesseract

Evaluation Metrics:

- Character Error Rate (CER)
- Word Error Rate (WER)
- Accuracy
- Precision
- Recall
- F1 Score
- Confidence Score
- Processing Time

```python
import os
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from jiwer import wer
from rapidfuzz.distance import Levenshtein

plt.style.use("ggplot")
```

```python
GROUND_TRUTH = "../data/samples/expected_output"

OCR_RESULTS = "../outputs/json"
```

```python
def character_error_rate(reference, prediction):

    if len(reference) == 0:
        return 0

    distance = Levenshtein.distance(
        reference,
        prediction
    )

    return distance / len(reference)
```

```python
def accuracy(reference, prediction):

    if len(reference) == 0:
        return 0

    matches = sum(
        r == p
        for r, p in zip(reference, prediction)
    )

    return matches / len(reference)
```

```python
reference = "Hello World"

prediction = "Hello Wor1d"

print("CER :", character_error_rate(reference, prediction))

print("WER :", wer(reference, prediction))

print("Accuracy :", accuracy(reference, prediction))
```

```python
results = []

for file in os.listdir(OCR_RESULTS):

    if file.endswith(".json"):

        path = os.path.join(
            OCR_RESULTS,
            file
        )

        df = pd.read_json(path)

        results.append(df)
```

```python
if results:

    data = pd.concat(
        results,
        ignore_index=True
    )

else:

    data = pd.DataFrame()

data.head()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    data["confidence"],
    bins=20
)

plt.title("Confidence Distribution")

plt.xlabel("Confidence")

plt.ylabel("Count")

plt.show()
```

```python
engine_scores = data.groupby(
    "engine"
)["confidence"].mean()

engine_scores
```

```python
engine_scores.plot.bar()

plt.ylabel("Average Confidence")

plt.title("OCR Engine Comparison")

plt.show()
```

```python
time_stats = data.groupby(
    "engine"
)["processing_time"].mean()

time_stats
```

```python
time_stats.plot.bar()

plt.ylabel("Seconds")

plt.title("Average Processing Time")

plt.show()
```

```python
cer_scores = data.groupby(
    "engine"
)["cer"].mean()

cer_scores
```

```python
cer_scores.plot.bar()

plt.title("Character Error Rate")

plt.ylabel("CER")

plt.show()
```

```python
wer_scores = data.groupby(
    "engine"
)["wer"].mean()

wer_scores
```

```python
wer_scores.plot.bar()

plt.ylabel("WER")

plt.show()
```

```python
best = engine_scores.idxmax()

print("Best Engine :", best)
```
