```python
import os
import random

import cv2
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image

plt.rcParams["figure.figsize"] = (12,8)
```

```python
IAM_PATH = "../data/iam"
CVL_PATH = "../data/cvl"
EMNIST_PATH = "../data/emnist"
CUSTOM_PATH = "../data/custom"
```

```python
datasets = {
    "IAM": IAM_PATH,
    "CVL": CVL_PATH,
    "EMNIST": EMNIST_PATH,
    "CUSTOM": CUSTOM_PATH
}

for name, path in datasets.items():

    print("="*40)

    print(name)

    print(path)

    print("Exists:", os.path.exists(path))
```

```python
IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tiff",
    ".webp"
)

def count_images(folder):

    total = 0

    for root, _, files in os.walk(folder):

        for file in files:

            if file.lower().endswith(IMAGE_EXTENSIONS):
                total += 1

    return total
```

```python
for dataset in datasets:

    total = count_images(datasets[dataset])

    print(f"{dataset}: {total} images")
```

```python
def random_image(folder):

    images = []

    for root, _, files in os.walk(folder):

        for file in files:

            if file.lower().endswith(IMAGE_EXTENSIONS):

                images.append(os.path.join(root,file))

    if not images:
        return None

    return random.choice(images)
```

```python
image_path = random_image(IAM_PATH)

print(image_path)

image = cv2.imread(image_path)

image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)

plt.imshow(image)

plt.axis("off")
```

```python
print("Height :",image.shape[0])

print("Width :",image.shape[1])

print("Channels :",image.shape[2])
```

```python
gray = cv2.cvtColor(image,cv2.COLOR_RGB2GRAY)

plt.imshow(gray,cmap="gray")

plt.axis("off")
```

```python
plt.hist(gray.ravel(),256)

plt.title("Pixel Distribution")

plt.show()
```

```python
_,thresh = cv2.threshold(

    gray,

    0,

    255,

    cv2.THRESH_BINARY+cv2.THRESH_OTSU

)

plt.imshow(thresh,cmap="gray")

plt.axis("off")
```

```python
blur = cv2.GaussianBlur(

    gray,

    (5,5),

    0

)

plt.imshow(

    blur,

    cmap="gray"

)

plt.axis("off")
```

```python
edges = cv2.Canny(

    gray,

    100,

    200

)

plt.imshow(

    edges,

    cmap="gray"

)

plt.axis("off")
```

```python
resized = cv2.resize(

    image,

    (800,600)

)

plt.imshow(resized)

plt.axis("off")
```

```python
fig = plt.figure(figsize=(15,10))

for i in range(6):

    img_path = random_image(IAM_PATH)

    img = cv2.imread(img_path)

    img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

    plt.subplot(2,3,i+1)

    plt.imshow(img)

    plt.axis("off")
```

```python
print("Mean Pixel :",gray.mean())

print("Std Dev :",gray.std())

print("Min :",gray.min())

print("Max :",gray.max())
```

```python
os.makedirs(

    "../data/processed",

    exist_ok=True

)

cv2.imwrite(

    "../data/processed/example.png",

    thresh

)

print("Saved.")
```
